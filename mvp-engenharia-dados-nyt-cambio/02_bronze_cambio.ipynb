```python
import yfinance as yf

dxy = yf.download("DX-Y.NYB", start="2025-01-01", end="2026-09-13").reset_index()
usdbrl = yf.download("BRL=X", start="2025-01-01", end="2026-09-13").reset_index()

dxy.columns = ['_'.join(col).strip('_') if isinstance(col, tuple) else col for col in dxy.columns]
usdbrl.columns = ['_'.join(col).strip('_') if isinstance(col, tuple) else col for col in usdbrl.columns]

df_dxy = spark.createDataFrame(dxy)
df_usdbrl = spark.createDataFrame(usdbrl)

print(f"DXY: {df_dxy.count()} linhas")
print(f"USDBRL: {df_usdbrl.count()} linhas")

df_dxy.write.mode("overwrite").saveAsTable("workspace.default.bronze_dxy")
df_usdbrl.write.mode("overwrite").saveAsTable("workspace.default.bronze_usdbrl")
```
